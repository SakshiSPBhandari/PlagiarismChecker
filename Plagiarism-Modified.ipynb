```python
import os
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import numpy as np
import pandas as pd  # Import Pandas for DataFrame

# Download NLTK resources (only need to do this once)
nltk.download('punkt')
nltk.download('stopwords')

def preprocess_text(text):
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    text = text.lower()
    tokens = word_tokenize(text)
    stop_words = set(stopwords.words('english'))
    stemmer = PorterStemmer()
    filtered_tokens = [stemmer.stem(token) for token in tokens if token not in stop_words]
    return ' '.join(filtered_tokens)

def vectorize(text):
    tfidf_vectorizer = TfidfVectorizer()
    tfidf_matrix = tfidf_vectorizer.fit_transform(text)
    return tfidf_matrix.toarray(), tfidf_vectorizer

def similarity(doc1, doc2):
    return cosine_similarity(doc1.reshape(1, -1), doc2.reshape(1, -1))[0][0]

def check_plagiarism(new_file_path, existing_files, tfidf_vectorizer):
    with open(new_file_path, 'r', encoding='utf-8') as new_file_contents:
        new_file_content = new_file_contents.read()
        preprocessed_new_file_content = preprocess_text(new_file_content)
        new_vector = tfidf_vectorizer.transform([preprocessed_new_file_content]).toarray()
        similarity_scores = []  # List to store similarity scores
        for file_name, text_vector in existing_files:
            sim_score = similarity(new_vector, text_vector)
            similarity_scores.append(sim_score)
            file_pair = sorted((os.path.basename(new_file_path), file_name))
            score = (file_pair[0], file_pair[1], sim_score)
            plagiarism_results.add(score)
        avg_similarity = np.mean(similarity_scores)  # Calculate average similarity
        return avg_similarity

directory_path = 'C:/Users/Sakshi/Desktop/DATASETS/TEXT FORMAT'

# List files with the .txt extension
ieee_files = [doc for doc in os.listdir(directory_path) if doc.endswith('.txt')]

# Initialize a list to store the contents of each file
ieee_data = []

for file in ieee_files:
    file_path = os.path.join(directory_path, file)
    try:
        with open(file_path, 'r', encoding='utf-8') as file_contents:
            content = file_contents.read()
            ieee_data.append(content)
    except FileNotFoundError:
        print(f"File not found: {file_path}")
    except PermissionError:
        print(f"Permission denied for file: {file_path}")
    except Exception as e:
        print(f"An error occurred while reading {file_path}: {str(e)}")

# Preprocess and vectorize existing data
preprocessed_ieee_data = [preprocess_text(text) for text in ieee_data]
vectors, tfidf_vectorizer = vectorize(preprocessed_ieee_data)
existing_files = list(zip(ieee_files, vectors))

# Example usage:
new_file_path = 'C:/Users/Sakshi/Desktop/DATASETS/TEXT FORMAT/same.txt'  # Replace with the actual path to the new file
avg_similarity = check_plagiarism(new_file_path, existing_files, tfidf_vectorizer)

# Create a DataFrame to store results
results_df = pd.DataFrame(columns=['File1', 'File2', 'Average Similarity'])
results_df.loc[0] = [os.path.basename(new_file_path), 'Existing Files', avg_similarity]

# Print the results DataFrame
print(results_df)
```
